```python
board=[['#']*6]*6
def check_win(board): #1. win    0.tie/inconclusive    -1:lose
  for i in range(3):
    if board[i][0]==board[i][1]==board[i][2]:
      if board[i][0]=='x':return 1
      elif board[i][0]=='o':return -1
    elif board[0][i]==board[1][i]==board[2][i]:
      if board[0][i]=='x':return 1
      elif board[0][i]=='o':return -1
  if board[0][0]==board[1][1]==board[2][2]:
    if board[0][0]=='x':return 1
    elif board[0][0]=='o':return -1
  return 0
#----------------------------------------------------
def display_board(board):
  print()
  for i in range(3):
    print()
    for j in range(3):
      print(board[i][j],end=' ')
#----------------------------------------------------
def min(board,display): # O's turn
  score=check_win(board)
  if score!=0:
    return [score,score]
  score=[2,-2]
  minimum=2
  maximum=-2
  minIndex=[-1,-1]
  maxIndex=[-1,-1]
  for i in range(3):
    for j in range(3):
      if board[i][j]=='#':
        board[i][j]='o'
        score=max(board,False) # X's turn
        board[i][j]='#'
      if minimum>score[0]:
        minimum=score[0]
        minIndex=[i,j]
      if maximum<score[1]:
        maximum=score[1]
        maxIndex=[i,j]
  if display==True:
    board[minIndex[0]][minIndex[1]]='o'
    display_board(board)
    return max(board,True)
  return [minimum,maximum]


#----------------------------------------------------
def max(board,display): # X's turn
  score=check_win(board)
  if score!=0:
    return [score,score]
  score=[2,-2]
  minimum=2
  maximum=-2
  minIndex=[-1,-1]
  maxIndex=[-1,-1]
  for i in range(3):
    for j in range(3):
      if board[i][j]=='#':
        board[i][j]='x'
        score=min(board,False) # O's turn
        board[i][j]='#'
      if minimum>score[0]:
        minimum=score[0]
        minIndex=[i,j]
      if maximum<score[1]:
        maximum=score[1]
        maxIndex=[i,j]
  if display==True:
    board[maxIndex[0]][maxIndex[1]]='x'
    display_board(board)
    return min(board,True)
  return [minimum,maximum]
#----------------------------------------------------
def minimax():
  board=[]*3
  for i in range(3):
    row=[]
    for j in range(3):
      row.append('#')
    board.append(row)
  score=max(board,True)
  if score[0]==1:
    print("MAX WINS")
  else:
    print("MIN WINS")

minimax()
```

```python
numRows=6
numCols=7
Graph=['#']*42
def return_index(graph,i,j):
  if i>=numRows or j>=numCols or i<0 or j<0:
    return '#'
  return graph[i*6+j]

def isFilled(graph,column):
  for i in range(numRows):
    if return_index(graph,i,column)=='#':
      return False
  return True
def insert_into(graph,column,colour):
  for i in range(numRows):
    if return_index(graph,i,column)=='#':
      graph[i*6+column]=colour
      return graph
def insert_outof(graph,column,colour):
  for i in range(numRows):
    if return_index(graph,i,column)!='#':
      graph[i*6+column]='#'
      return graph
    
def check_result(graph):
  for i in range(6):
    for j in range(7):
      equals=False;
      if return_index(graph,i,j)==return_index(graph,i,j+1)==return_index(graph,i,j+2)==return_index(graph,i,j+3):
        equals=True;
      elif return_index(graph,i,j)==return_index(graph,i+1,j)==return_index(graph,i+2,j)==return_index(graph,i+3,j):
        equals=True;
      elif return_index(graph,i,j)==return_index(graph,i+1,j+1)==return_index(graph,i+2,j+2)==return_index(graph,i+3,j+3):
        equals=True;
      elif return_index(graph,i,j)==return_index(graph,i+1,j-1)==return_index(graph,i+2,j-2)==return_index(graph,i+3,j-3):
        equals=True;
      if equals==True:
        if return_index(graph,i,j)=='g':
          return 1
        elif return_index(graph,i,j)=='b':
          return -1
  return 0
def display(graph):
  print("\n\n")
  for i in range(6):
    print()
    for j in range(7):
      print(return_index(graph,i,j),end=' ')

def min(graph,displ,depth):
  if depth>=7 and displ==False: return 0
  #print("depth:",depth)
  if displ==True:
    display(graph)
  result=check_result(graph)
  if result!=0:
    return result
  filled=True
  minIndex=-1
  minValue=1
  for col in range(numCols):
    if not isFilled(graph,col):
      filled=False
      insert_into(graph,col,'b')
      value=max(graph,False,depth+1);
      insert_outof(graph,col,'b')
      if value<minValue: 
        minValue=value
        minIndex=col
  if filled==True:
    return result
  elif filled==False:
    if displ==False:
      return value
    else:
      return max(insert_into(graph,minValue,'b'),True,depth+1)
    
def max(graph,displ,depth):
  if depth>=7 and displ==False: return 0
  #print("depth:",depth)
  if displ==True:
    display(graph)
  result=check_result(graph)
  if result!=0:
    return result
  filled=True
  maxIndex=-1
  maxValue=1
  for col in range(numCols):
    if not isFilled(graph,col):
      filled=False
      insert_into(graph,col,'g')
      value=min(graph,False,depth+1);
      insert_outof(graph,col,'g')
      if value>maxValue: 
        maxValue=value
        maxIndex=col
  if filled==True:
    return result
  elif filled==False:
    if displ==False:
      return value
    else:
      return min(insert_into(graph,maxValue,'g'),True,depth+1)
  
if max(Graph,True,0)==1: print("Max wins")
else: print("Min Wins")


  







```
