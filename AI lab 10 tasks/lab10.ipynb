Part 1: loan value

```python
bank_data_clean=bank_data.drop(columns=['user_id','date_of_birth','address','email','inquiry_purpose_code','asset_class_cd','asset_code'])
bank_data_clean
```

```python

bank_data_clean=bank_data_clean.dropna()
bank_data_clean
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
x_data=pd.get_dummies(bank_data_clean.drop(columns=['approved_loan_amount','approved']))
y_data=bank_data_clean['approved_loan_amount']
x_train, x_test, y_train,y_test= train_test_split(x_data,y_data)
model=LinearRegression().fit(x_train,y_train)
```

```python
import numpy as np
y_test_pred=np.array(model.predict(x_test))
y_test=np.array(y_test)

```

```python
def get_accuracy(actual,predicted):
    closeness=abs(1-abs(actual-predicted)/(abs(actual+predicted)/2))
    return float(np.mean(closeness))

```

```python
get_accuracy(y_test,y_test_pred)
```
